```python
# Exercise 1: Handling ZeroDivisionError

def divide_numbers(numerator, denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by Zero.")
        return None


numerator = float(input("Enter the numerator: "))
denominator = float(input("Enter the denominator: "))

result = divide_numbers(numerator, denominator)

if result is not None:
    print("Result:", result)
```

```python
# Exercise 2: Handling ValueError

import math

def calculate_square_root(number1):
    try:
        if number1 < 0:
            raise ValueError
        return math.sqrt(number1)
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float value.")
        return None


number1 = float(input("Enter a number: "))

result = calculate_square_root(number1)

if result is not None:
    print("Square root:", result)
```

```python
# Exercise 3: Handling Generic Exceptions

def calculate_result(num):
    try:
        result = num / (num - 5)
        print("Result:", result)
    except Exception as e:
        print("An error occurred during calculation.")


num = float(input("Enter a number: "))

calculate_result(num)
```
