```python
# Problem 1: Fixing the buggy code using super()

class Person:
    
    
    def __init__(self, name, id):
        self.name = name
        self.id = id
    
    
    def Display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):
    
   
    def __init__(self, name, id):
        super().__init__(name, id)
    
    def Print(self):
        print("Emp class called")


Emp_details = Emp("Mayank", 103)


Emp_details.Display()


Emp_details.Print()
```

```python
# Problem 2: Program to check whether two strings are anagrams

def anagram_check(str1, str2):
    
    
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    
    
    if sorted(str1) == sorted(str2):
        return True
    else:
        return False


string1 = input("Enter first string: ")
string2 = input("Enter second string: ")

if anagram_check(string1, string2):
    print("The strings are Anagrams.")
else:
    print("The strings are not Anagrams.")
```

```python
# Problem 3: Program to check Armstrong Number

num = int(input("Enter a number: "))


original_num = num


digits = len(str(num))

sum_of_powers = 0


while num > 0:
    digit = num % 10
    sum_of_powers += digit ** digits
    num = num // 10


if sum_of_powers == original_num:
    print(original_num, "is an Armstrong Number.")
else:
    print(original_num, "is not an Armstrong Number.")
```

```python
# Problem 4: Deal with Classes and Objects
# Scenario: Car Dealership's Inventory Management System

class Vehicle:
    
    #
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "white"
        self.seating_capacity = 0
    
    
    def assign_seating_capacity(self, seating_capacity):
        self.seating_capacity = seating_capacity
    
   
    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)



vehicle1 = Vehicle(200, 20)
vehicle1.assign_seating_capacity(5)


vehicle2 = Vehicle(180, 25)
vehicle2.assign_seating_capacity(4)
```

```python
print("Vehicle 1 Details")
print("------------------")

vehicle1.display_properties()
```

```python
print("Vehicle 2 Details")
print("------------------")

vehicle2.display_properties()
```
