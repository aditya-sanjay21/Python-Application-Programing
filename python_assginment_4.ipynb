```python
players = [
    ["Serena Williams", "Tennis", 23],
    ["Lionel Messi", "Soccer", 7],
    ["Michael Phelps", "Swimming", 23],
    ["Usain Bolt", "Athletics", 8],
    ["Roger Federer", "Tennis", 20],
    ["Cristiano Ronaldo", "Soccer", 5]
]
```

```python
for player in players:
    if player[0] == "Lionel Messi":
        if player[2] > 10:
            print(player[0])
            print(player[1])
            print(player[2])
        else:
            print("Does not have more than 10 achievements.")
```

```python
for player in players:
    if player[1] == "Tennis" or player[2] == 20:
        print(player[0], "meets the condition")
```

```python
for player in players:
    if player[2] < 10 and player[1] != "Soccer":
        print(player[0])
        print(player[1])
        print(player[2])
```

```python
num = int(input("Enter an integer from 1 to 9999: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")

if str(num) == str(num)[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")

digit_sum = sum(int(digit) for digit in str(num))

if num % digit_sum == 0:
    print("Divisible by the sum of its digits")
else:
    print("Not divisible by the sum of its digits")
```

```python
animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(animals):
    if len(animals[i]) == 7:
        seven_letter_animals.append(animals[i])
    i += 1

print(seven_letter_animals)
```
